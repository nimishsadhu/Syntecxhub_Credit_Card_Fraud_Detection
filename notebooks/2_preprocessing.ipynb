# Preprocessing

Remove exact duplicate rows and keep the cleaned data for the next notebooks.

```python
from pathlib import Path
import pandas as pd

df = pd.read_csv("../data/creditcard.csv")
print("Before:", df.shape, "duplicates:", int(df.duplicated().sum()))
```

```python
cleaned_df = df.drop_duplicates().copy()
print("After:", cleaned_df.shape, "duplicates:", int(cleaned_df.duplicated().sum()))
print("Missing values:", int(cleaned_df.isna().sum().sum()))
print("Class counts:")
print(cleaned_df["Class"].value_counts().sort_index())
```

```python
project_root = Path.cwd().parent 
data_dir = project_root/ "data"
data_dir.mkdir(exist_ok=True)
cleaned_path = data_dir / "cleaned_data.csv.gz"
cleaned_df.to_csv(cleaned_path, index=False, compression="gzip")
print("Saved:", cleaned_path.resolve())
```

Duplicates were exact row matches, so I removed them. The remaining data has no missing values and keeps all original feature columns.
